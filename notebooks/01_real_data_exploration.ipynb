# Real notification episode exploration
Run the pipeline audit/features first. This notebook describes selected observed episodes, not held-out model evaluation. Removal is not reading.

```python
from pathlib import Path
import pandas as pd
root = Path.cwd() if Path("data").exists() else Path.cwd().parent
episodes = pd.read_csv(root / "data/processed/episodes.csv", dtype={"participant": str})
episodes.groupby("participant").duration_seconds.describe()
```
